```python
# ham sipariş verisi
orders = [
    {
        "customer":"Ahmet",
        "items": ["Laptop","Mouse","Mouse"],
        "total":18500,
        "is_paid":True
    },
    {
        "customer":"Zeynep",
        "items": ["Keyboard"],
        "total":2500,
        "is_paid":False
    },
    {
        "customer":"Mehmet",
        "items": ["Monitor","HDMI Cable","Monitor"],
        "total":7200,
        "is_paid":True
    },
    {
        "customer":"Mehmet",
        "items": ["Monitor","HDMI Cable","Monitor"],
        "total":7200,
        "is_paid":True
    }
]
```

```python
#toplam sipariş sayısı
total_count = len(orders)

#başlangıç değerleri
paid_orders = 0 
unpaid_orders = 0 

# 1.sipariş
if orders[0]["is_paid"]:
    paid_orders += 1
else: 
    unpaid_orders += 1

# 2.sipariş
if orders[1]["is_paid"]:
    paid_orders += 1
else:
    unpaid_orders += 1

# 3.sipariş
if orders[2]["is_paid"]:
    paid_orders += 1
else:
    unpaid_orders += 1

# 4.sipariş
if orders[3]["is_paid"]:
    paid_orders += 1
else:
    unpaid_orders += 1
```

```python
#önce tüm isimleri bir listeye alalım
all_customers = [orders[0]["customer"],orders[1]["customer"],orders[2]["customer"],orders[3]["customer"]]

#set kullanarak tekrarları temizle
unique_customers_set = set(all_customers)
```

```python
#liste metodları ve toplama ile tüm ürünleri birleştir
all_products = orders[0]["items"] + orders[1]["items"] + orders[2]["items"] + orders[3]["items"] 

#unique ürünleri set ile ayır
unique_products = set(all_products)
```

```python
# 1.Sipariş Durumu
if orders[0]["is_paid"]:
    status0 = "Sipariş onaylandı."
else:
    status0 = "Ödeme bekleniyor."
    
# 2.Sipariş Durumu
if orders[1]["is_paid"]:
    status1 = "Sipariş onaylandı."
else:
    status1 = "Ödeme bekleniyor."
    
# 3.Sipariş Durumu
if orders[2]["is_paid"]:
    status2 = "Sipariş onaylandı."
else:
    status2 = "Ödeme bekleniyor."
    
# 4.Sipariş Durumu
if orders[3]["is_paid"]:
    status3 = "Sipariş onaylandı."
else:
    status3 = "Ödeme bekleniyor."
```

```python
# Her sipariş için tuple oluşturma
summary0 = (orders[0]["customer"], orders[0]["total"])
summary1= (orders[1]["customer"], orders[1]["total"])
summary2= (orders[2]["customer"], orders[2]["total"])
summary3 = (orders[3]["customer"], orders[3]["total"])

all_summaries = [summary0, summary1, summary2, summary3]

# Tuple neden değiştirilemez?
# CEVAP: Tuple'lar 'immutable' (değiştirilemez) koleksiyonlardır. 
# Sipariş tutarı gibi kritik verilerin yanlışlıkla güncellenmesini (örn: summary0[1] = 0) engellemek için tercih edilir.
```

```python
print("Toplam sipariş:", total_count)
print("Ödenmiş sipariş:", paid_orders)
print("Ödenmemiş sipariş:", unpaid_orders)

print("\nUnique ürünler:", unique_products)
print("Sipariş özetleri:", all_summaries)
```
